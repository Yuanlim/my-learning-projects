```python
print("fixed 10 size stack");
CoursesJourney: fixedStack[str] = fixedStack[T](maxElement=10) 
Journey:List[str] = ["Computer programming", "Web development", "Data structure", "Algorithm", "Assembly Language", "Android Apps Labatory", "System Design and Analysis", "Computer Graphic", "Digital Image Processing", "Computer Vision"];

print("10 member inserted");
for i in range(len(Journey)):
  CoursesJourney.push(Journey[i]);

print(f"IsFull: {CoursesJourney.isFull()}");
print(f"IsEmpty: {CoursesJourney.isEmpty()}");
# CoursesJourney.Push("Quantum Computing"); # Error exceed max

# Pop until empty
if not(CoursesJourney.isEmpty()) :
   print(CoursesJourney.pop(), end=" ")
while not(CoursesJourney.isEmpty()):
  print(f"←-- {CoursesJourney.pop()}", end=" ");
print();
```
